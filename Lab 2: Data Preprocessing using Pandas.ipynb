# Lab 2: Data Preprocessing using Pandas

### Objectives:
1. Import data and perform exploratory checks.
2. Handle missing values using mean and mode imputation.
3. Remove duplicate records from the dataset.
4. Encode categorical variables into numerical formats.
5. Apply Normalization and Standardization.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder

df = pd.read_csv("/content/synthetic_dataset.csv")
```

```python
df.info()
df.shape
```

## 2. Removing Duplicates :
Duplicate rows are repeated records. Removing them avoids counting the same data twice.

```python
df = df.drop_duplicates()
print(f"Dataset shape after removing duplicates: {df.shape}")
```

## 3. Checking Missing Values :
This counts the empty (NaN) cells in each column.

```python
df.isna().sum()
```

## 4. Handling Missing Values :
Numerical columns are filled with the mean. Categorical columns are filled with the mode (most frequent value).

```python
for col in ["Price", "Rating", "Discount"]:
    df[col] = df[col].fillna(df[col].mean())

for col in ["Category", "Stock"]:
    df[col] = df[col].fillna(df[col].mode()[0])

df.isna().sum()
```

## 5. Encoding Categorical Variables :
Machine learning models need numbers, so text categories are converted to numeric labels using Label Encoding.

```python
le = LabelEncoder()
df["Category"] = le.fit_transform(df["Category"])
df["Stock"] = le.fit_transform(df["Stock"])
df.head()
```

## 6. Standardization :
Standardization rescales values to mean 0 and standard deviation 1 using z = (x - mean) / std.

```python
df_std = df.copy()
df_std[["Price", "Rating", "Discount"]] = StandardScaler().fit_transform(df[["Price", "Rating", "Discount"]])
df_std.head()
```

## 7. Normalization :
Normalization (Min-Max scaling) rescales values to the range 0 to 1 using (x - min) / (max - min).

```python
df_norm = df.copy()
df_norm[["Price", "Rating", "Discount"]] = MinMaxScaler().fit_transform(df[["Price", "Rating", "Discount"]])
df_norm.head()
```
